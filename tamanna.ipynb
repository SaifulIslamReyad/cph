```python
import matplotlib.pyplot as plt
import numpy as np

# Create figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Given points (Quantity, Price)
points = {
    'A': (5, 100),
    'B': (10, 60),
    'C': (15, 40),
    'D': (20, 20),
    'E': (25, 10)
}

# Extract x and y values
quantities = [point[0] for point in points.values()]
prices = [point[1] for point in points.values()]

# Simple line plot connecting the points (less smooth but doesn't need scipy)
ax.plot(quantities, prices, color='blue', linewidth=2, label='Demand Curve', marker='o')

# Label the points
for label, (x, y) in points.items():
    ax.text(x+0.5, y+3, label, fontsize=12, ha='center', va='bottom')

# Set labels and title
ax.set_xlabel('Quantity', fontsize=12)
ax.set_ylabel('Price', fontsize=12)
ax.set_title("An individual's Demand Curve for Shifts", fontsize=14, pad=20)

# Set the limits
ax.set_xlim(0, 30)
ax.set_ylim(0, 110)

# Add grid
ax.grid(True, linestyle='--', alpha=0.6)

# Show the plot
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample data similar to the image (Price vs Quantity)
quantity = [30, 60, 90, 120, 180, 210]
price = [10, 8, 6, 4, 2, 1]  # You can adjust these to match the actual demand schedule if needed

# Plotting the Market Demand Curve
plt.figure(figsize=(8, 6))
plt.plot(quantity, price, marker='o', linestyle='-', color='black', label='Market Demand Curve For a Commodity')

# Labeling the points D and D' (adjusted positions)
plt.text(quantity[0] + 4, price[0] - 0.1, 'D', fontsize=12)     # Label beside the first point
plt.text(quantity[-1] + 2, price[-1] - 0.3, "D'", fontsize=12)  # Label beside the last point

# Adding titles and labels
plt.title('Market Demand Curve', fontsize=14)
plt.xlabel('Quantity', fontsize=12)
plt.ylabel('Price', fontsize=12)

# Grid and legend
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Set up the plot
plt.figure(figsize=(8, 6))

# Plot the demand curve (sloping down)
plt.plot([0, 8], [10, 0], color='black')  # Main demand curve

# Price levels: P1 and P2
plt.hlines(7, 0, 6, colors='black', linestyles='-')  # P1
plt.hlines(4, 0, 7, colors='black', linestyles='-')  # P2

# Vertical lines for M1, M2, M3
plt.vlines(3, 0, 10, colors='black', linestyles='-')  # M1
plt.vlines(5, 0, 10, colors='black', linestyles='-')  # M2
plt.vlines(7, 0, 10, colors='black', linestyles='-')  # M3

# Label Points a and b
plt.text(3, 7.2, 'a', fontsize=12)
plt.text(7, 3.8, 'b', fontsize=12)

# Arrows for contraction and extension
plt.annotate('',
             xy=(5, 7), xytext=(3, 7),
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.annotate('',
             xy=(5, 4), xytext=(7, 4),
             arrowprops=dict(facecolor='black', arrowstyle='->'))

# Labels for arrows
plt.text(5.2, 7.2, 'Contraction in demand', fontsize=11)
plt.text(5.2, 4.2, 'Extension in demand', fontsize=11)

# Labels for axes
plt.xlabel('Demand', fontsize=12)
plt.ylabel('Price', fontsize=12)

# X-axis custom labels
plt.xticks([0, 3, 5, 7], ['0', 'M1', 'M2', 'M3'])

# Y-axis custom labels
plt.yticks([4, 7, 10], ['P2', 'P1', 'P'])

# Remove top and right borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Add grid for better readability
plt.grid(False)
plt.title('Contraction and Extension in Demand', fontsize=14)

# Show the plot
plt.show()
```
