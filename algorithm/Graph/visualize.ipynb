```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a graph
G = nx.Graph()

# List of edges with weights
edges =  [(1, 2, 1), (2, 4, 1), (2, 5, 1), (4, 6, 1), (6, 3, 1), (3, 7, 1)]

# Add edges to the graph
for u, v, w in edges:
    G.add_edge(u, v, weight=w)

# Define positions for nodes
pos = nx.spring_layout(G, seed=42, k=0.3)  # Adjust k for better spacing

# Draw the graph
plt.figure(figsize=(12, 7))
nx.draw(G, pos, with_labels=True, node_size=1200, node_color="skyblue", 
        edge_color="gray", font_size=12, font_weight="bold", width=2)

# Draw edge labels (weights)
edge_weights = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_weights, font_size=10, font_color="red")

# Show the graph
plt.title("Graph Visualization with Weights", fontsize=14)
plt.show()
```
